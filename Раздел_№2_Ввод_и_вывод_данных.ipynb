## **Раздел №2. Вывод и ввод данных**

Прежде чем начать работать с типами данных, важно научиться выводить информацию на экран и получать ввод от пользователя.

#### Функция **print()**

Функция `print()` используется для вывода информации на экран. Вы можете выводить текст, числа и результаты вычислений.

**Пример использования:**

```python
print("Привет, мир!")
```

**Зачем это нужно?**

С помощью `print()` вы можете сообщать пользователю результаты работы программы или просто отображать необходимую информацию.

#### Функция **input()**

Функция `input()` используется для получения данных от пользователя. Она останавливает выполнение программы и ожидает, пока пользователь введёт данные и нажмёт Enter. Результат введенных данных всегда является строкой.

**Пример использования:**

```python
name = input("Введите ваше имя: ")
print("Привет, " + name + "!")
```

```python
print(print())
```

**Зачем это нужно?**

С помощью `input()` вы можете создавать интерактивные программы, которые реагируют на действия пользователя.
