Вася составляет 6-буквенные коды из букв Б, А, Н, К, И, Р. Каждая допустимая гласная буква может входить в код не более одного раза. Сколько кодов может составить Вася?

```python
cnt = 0
for pos1 in 'BANKIR':
  for pos2 in 'BANKIR':
    for pos3 in 'BANKIR':
      for pos4 in 'BANKIR':
        for pos5 in 'BANKIR':
          for pos6 in 'BANKIR':
            s = pos1 + pos2 + pos3 + pos4 + pos5 + pos6
            if s.count('A') <= 1 and s.count('I') <= 1:
              cnt += 1
print(cnt)
```

```python
from itertools import product

cnt = 0
for i in product('BANKIR', repeat=6):
  if i.count('A') <= 1 and i.count('I') <= 1:
    cnt += 1

print(cnt)
```

Вася составляет 4-буквенные слова, в которых могут быть использованы только буквы В, И, Ш, Н, Я. Слова должны содержать хотя бы одну букву В. Каждая из других допустимых букв может встречаться в слове любое количество раз или не встречаться совсем. Слово не должно начинаться с буквы Ш и должно оканчиваться не гласными буквами. Словом считается любая допустимая последовательность букв, не обязательно осмысленная. Сколько существует таких слов, которые может написать Вася?

```python
from itertools import product

cnt = 0
for i in product('VISNY', repeat=4):
  if i.count('V') >= 1 and i.count('I') >= 1 and i[0] == 'S' and i[3] in 'VSN':
    cnt += 1

print(cnt)
```

```python
from itertools import product

cnt = 0
for i in product('VISNY', repeat=4):
  if i.count('V') >= 1 and i.count('I') >= 1 and i[0] == 'S' and i[3] in 'VSN':
    cnt += 1

print(cnt)
```

Вася составляет 5-буквенные слова, в которых есть только буквы П, А, У, К, причём буквы К не могут стоять рядом друг с другом (запрещено сочетание букв КК). Каждая из допустимых букв может встречаться в слове любое количество раз или не встречаться совсем. Словом считается любая допустимая последовательность букв, не обязательно осмысленная. Сколько существует таких слов, которые может написать Вася?

```python
cnt = 0
for pos1 in 'PAYK':
  for pos2 in 'PAYK':
    for pos3 in 'PAYK':
      for pos4 in 'PAYK':
        for pos5 in 'PAYK':
          s = pos1 + pos2 + pos3 + pos4 + pos5
          if 'KK' not in s:
            cnt += 1
print(cnt)
```

```python
from itertools import product

cnt = 0
for i in product('PAYK', repeat=5):
  s = ''.join(i)
  if 'KK' not in s:
    cnt += 1
print(cnt)
```

Определите количество четырёхзначных чисел, записанных в восьмеричной системе счисления, которые начинаются с четных цифр, не оканчиваются на цифры, меньшие 5, а также содержат в своей записи более двух цифр 6.

```python
cnt = 0
for pos1 in '1234567':
  for pos2 in '01234567':
    for pos3 in '01234567':
      for pos4 in '01234567':
        s = pos1 + pos2 + pos3 + pos4
        if s[0] in '246' and s[3] in '567' and s.count('6') > 2:
          cnt += 1
print(cnt)
```

```python
cnt = 0
for pos1 in '246':
  for pos2 in '01234567':
    for pos3 in '567':
      for pos4 in '01234567':
        s = pos1 + pos2 + pos3 + pos4
        if s.count('6') > 2:
          cnt += 1
print(cnt)
```

```python
from itertools import product

cnt = 0
for i in product('01234567', repeat = 4):
  if i[0] != '0' and i[0] in '246' and i[3] in '567' and i.count('6') > 2:
    cnt += 1
print(cnt)
```
